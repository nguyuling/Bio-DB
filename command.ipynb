### **OPERATORS**

#### **Comparison Operator**
- $eq - equal to
- $ne - not equal to
- $gt - greater than
- $gte - greater than / equal to
- $lt - less than
- $lte - less than / equal to

```python
db.<collection>.<operation>{(
    <field>: { 
        $comp_op_1: <value>,
        $comp_op_2: <value>
    }
)}
```

#### **Logical Oeprator**
- $all - match all, array contains all the specified values
- $in - match at least one, array contains at least one of the specified values
- $nor - do not match, array do not have all the specified value
- $size - size of array must match the specified value
- $elemMatch - array contains the exact element
- $not - do not match, field's value is not the specified value 

```python
db.<collection>.<operation>(
    {
        <array>: {
            $lg_op: ["item_1", "item_2"]
        }
    }
)
```

```python
db.<collection>.<operation>(
    {
        <array>: { $size: n }
    }
)
```

```python
db.<collection>.<operation>(
    {
        <array>: { 
            $elemMatch: { <field>: value }
        }
    }
)
```

```python
db.<collection>.<operator>({
    <field>: {
        $not: { $op: "value" }
    }
})

// where $op can be $eq, $gt, $lt...
```

### **AGGREGATION**

Match
- $match - find documents where the selected field matches the specified value
- $group - group  documents using a specified key (_id) and apply aggregation operators on the grouped data
- $project - select field to be shown
- $sort - sort any field by ascending or descending order
- $limit - restrict the number of documents returned
- $unwind - deconstruct an array

```python
db.<collection.aggregate([
    { 
        $match: { 
            "field_1": "value",
            "field_2": "value"
        }
    }
])
```

```python
db.<collection>.aggregate([
    {
        $group: {
            "_id": "$<field>",
            <new_field>: { $op: "$<field>" }
        }
    }
])

// where $op can be $sum, $avg, $min, $max
```

```python
db.<collection>.aggregate([
    {
        $project: {
            "field_1": 1,
            "field_2": 0
        }
    }
])
```

```python
db.<collection>.aggregate([
    {
        $sort: {
            "field_1": 1,
            "field_2": -1
        }
    }
])
```

```python
db.<collection>.aggregate([
    {
        $limit: n
    }
])
```

```python
db.<collection>.aggregate([
    {
        $unwind: <array>
    }
])
```

```python
db.restaurants.aggregate([
  {
    $unwind: "$address"
  },
  {
    $group: {
      "_id": "$address.street",
      total_restaurants: { $sum: 1 }
    }
  },
  {
    $sort: { total_restaurants: -1 }
  },
  {
    $limit: 3
  }
])
```
